```python
import os
import re

import numpy as np
```

## Ours

```python
train_score = {
    "FGARI": [], "ARI": [], "AMIO": [], "AMI": [],
    "mIoU": [], "PSNR": [], "SSIM": [], "LPIP": [], "FPS": []
}

with open("COLMAP/OCTA/ours/ours_extend_h200_train.txt") as train:
    for x in train:
        split = re.split(r"[:,=]", x)

        train_score["FGARI"].append(float(split[2]))
        train_score["ARI"].append(float(split[4]))
        train_score["AMIO"].append(float(split[6]))
        train_score["AMI"].append(float(split[8]))
        train_score["mIoU"].append(float(split[10]))
        train_score["PSNR"].append(float(split[12]))
        train_score["SSIM"].append(float(split[14]))
        train_score["LPIP"].append(float(split[16]))
        train_score["FPS"].append(float(split[18]))
results = {}

for k, v in train_score.items():
    mean = np.mean(v)
    std = np.std(v)  # sample std: use np.std(v, ddof=1) if needed
    results[k] = (mean, std)
print(results)
for k, (mean, std) in results.items():
    print(f"{k}: {mean:.3f} ± {std:.3f}")
```
